Config

```python
class ExperimentConfig:
    DATA_ROOT = Path("./data")
    ARTIFACTS_ROOT = Path("./outputs")
    SYNTH_IMAGES_DIR = ARTIFACTS_ROOT / "synthetic_images"
    CKPT_DIR = ARTIFACTS_ROOT / "checkpoints"
    METRICS_DIR = ARTIFACTS_ROOT / "results"

    DATASET_NAME = "CIFAR10"
    NUM_LABELS = 10
    IMBALANCE_FRACTION = 0.1
    MINORITY_CLASS_ID = 3
    NUM_SYNTH_IMAGES = 500

    BACKBONE_TYPE = "ResNet50"
    TRAIN_BATCH_SIZE = 64
    TRAIN_EPOCHS = 100
    BASE_LR = 0.001
    L2_WEIGHT_DECAY = 1e-4

    ENABLE_CONTROLNET = True
    CONTROLNET_MODE = "canny"
    SD_MODEL_ID = "runwayml/stable-diffusion-v1-5"
    CONTROLNET_MODEL_ID = "lllyasviel/sd-controlnet-canny"
    COMPUTE_DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    RANDOM_SEED = 42

    CFG_GUIDANCE_SCALE = 7.5
    DIFFUSION_STEPS = 50
    IMAGES_PER_PROMPT = 1

    CIFAR10_TEXT_PROMPTS = {
        0: "a high-quality image of an airplane, natural colors, realistic lighting",
        1: "a high-quality image of a car on a road, detailed, natural background",
        2: "a realistic image of a bird in the wild, detailed feathers, clear focus",
        3: "a detailed image of a cat, sharp focus, natural lighting, realistic colors",
        4: "a realistic image of a deer in nature, full body visible, clear background",
        5: "a high-quality image of a dog outdoors, detailed fur, natural setting",
        6: "a realistic image of a frog near water or on a leaf, vivid colors",
        7: "a detailed image of a horse, side view, clear background, natural light",
        8: "a realistic image of a ship on the water, visible details and reflections",
        9: "a high-quality image of a truck on a road, detailed, realistic environment",
    }

    def __init__(self):
        self.DATA_ROOT.mkdir(exist_ok=True)
        self.ARTIFACTS_ROOT.mkdir(exist_ok=True)
        self.SYNTH_IMAGES_DIR.mkdir(exist_ok=True)
        self.CKPT_DIR.mkdir(exist_ok=True)
        self.METRICS_DIR.mkdir(exist_ok=True)


config = ExperimentConfig()
torch.manual_seed(config.RANDOM_SEED)
np.random.seed(config.RANDOM_SEED)

```

load data

```python
class ImbalancedCIFAR10Dataset(Dataset):
    def __init__(
        self,
        train: bool = True,
        transform=None,
        imbalance_fraction: float = 1.0,
        minority_class_id: int | None = None,
    ):
        self.base_dataset = datasets.CIFAR10(
            root=config.DATA_ROOT,
            train=train,
            download=True,
            transform=transform,
        )
        self.sample_indices = self._build_imbalanced_index_list(
            imbalance_fraction=imbalance_fraction,
            minority_class_id=minority_class_id,
        )
        self.transform = transform

    def _build_imbalanced_index_list(
        self,
        imbalance_fraction: float,
        minority_class_id: int | None,
    ) -> list[int]:
        class_to_indices: dict[int, list[int]] = {}
        for sample_idx, (_, label) in enumerate(self.base_dataset):
            if label not in class_to_indices:
                class_to_indices[label] = []
            class_to_indices[label].append(sample_idx)

        if minority_class_id is not None:
            original_count = len(class_to_indices[minority_class_id])
            minority_count = int(original_count * imbalance_fraction)
            class_to_indices[minority_class_id] = class_to_indices[minority_class_id][:minority_count]

        index_buffer: list[int] = []
        for _, idx_list in class_to_indices.items():
            index_buffer.extend(idx_list)

        return index_buffer

    def __len__(self) -> int:
        return len(self.sample_indices)

    def __getitem__(self, index: int):
        mapped_idx = self.sample_indices[index]
        image, label = self.base_dataset[mapped_idx]
        return image, label


def build_data_loaders():
    transform_train = transforms.Compose([
        transforms.RandomCrop(32, padding=4),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
        transforms.Normalize(
            mean=(0.4914, 0.4822, 0.4465),
            std=(0.2470, 0.2430, 0.2610),
        ),
    ])

    transform_eval = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(
            mean=(0.4914, 0.4822, 0.4465),
            std=(0.2470, 0.2430, 0.2610),
        ),
    ])

    train_dataset_full = ImbalancedCIFAR10Dataset(
        train=True,
        transform=transform_train,
        imbalance_fraction=1.0,
        minority_class_id=None,
    )

    train_dataset_imbalanced = ImbalancedCIFAR10Dataset(
        train=True,
        transform=transform_train,
        imbalance_fraction=config.IMBALANCE_FRACTION,
        minority_class_id=config.MINORITY_CLASS_ID,
    )

    eval_dataset = datasets.CIFAR10(
        root=config.DATA_ROOT,
        train=False,
        download=True,
        transform=transform_eval,
    )

    train_loader_full = DataLoader(
        train_dataset_full,
        batch_size=config.TRAIN_BATCH_SIZE,
        shuffle=True,
        num_workers=8,
        pin_memory=True,
    )

    train_loader_imbalanced = DataLoader(
        train_dataset_imbalanced,
        batch_size=config.TRAIN_BATCH_SIZE,
        shuffle=True,
        num_workers=8,
        pin_memory=True,
    )

    eval_loader = DataLoader(
        eval_dataset,
        batch_size=config.TRAIN_BATCH_SIZE,
        shuffle=False,
        num_workers=8,
        pin_memory=True,
    )

    return {
        "train_full": train_loader_full,
        "train_imbalanced": train_loader_imbalanced,
        "test": eval_loader,
        "train_full_dataset": train_dataset_full,
    }
```

Image generator

```python
class SyntheticImageGenerator:
    def __init__(self, class_id: int, class_label: str):
        self.class_id = class_id
        self.class_label = class_label
        self.device = config.COMPUTE_DEVICE
        self.pipeline = None
        self._build_pipeline()

    def _build_pipeline(self) -> None:
        print(f"[Gen] Loading Stable Diffusion pipeline for class '{self.class_label}'...")

        if config.ENABLE_CONTROLNET:
            try:
                controlnet = ControlNetModel.from_pretrained(
                    config.CONTROLNET_MODEL_ID,
                    torch_dtype=torch.float16,
                    use_safetensors=True,
                )

                self.pipeline = StableDiffusionControlNetPipeline.from_pretrained(
                    config.SD_MODEL_ID,
                    controlnet=controlnet,
                    torch_dtype=torch.float16,
                    use_safetensors=True,
                )
            except Exception as exc:
                print(f"[Gen] Failed to load ControlNet ({exc}), falling back to base Stable Diffusion")
                self.pipeline = DiffusionPipeline.from_pretrained(
                    config.SD_MODEL_ID,
                    torch_dtype=torch.float16,
                    use_safetensors=True,
                )
        else:
            self.pipeline = DiffusionPipeline.from_pretrained(
                config.SD_MODEL_ID,
                torch_dtype=torch.float16,
                use_safetensors=True,
            )

        self.pipeline.to(self.device)
        self.pipeline.enable_attention_slicing()

        if hasattr(self.pipeline, "enable_sequential_cpu_offload"):
            self.pipeline.enable_sequential_cpu_offload()

    def _build_control_image(self, size: Tuple[int, int] = (512, 512)) -> Image.Image:
        if config.CONTROLNET_MODE == "canny":
            control_image = Image.new("RGB", size, color="white")
            drawer = ImageDraw.Draw(control_image)

            num_strokes = np.random.randint(3, 8)
            for _ in range(num_strokes):
                x1 = np.random.randint(0, size)
                y1 = np.random.randint(0, size)[1]
                x2 = np.random.randint(0, size)
                y2 = np.random.randint(0, size)[1]
                drawer.line([(x1, y1), (x2, y2)], fill="black", width=2)

            return control_image

        return Image.new("RGB", size, color="gray")

    def generate_batch(self, num_images: int = 10) -> List[Image.Image]:
        print(f"[Gen] Generating {num_images} synthetic samples for class '{self.class_label}'...")

        prompt = config.CIFAR10_TEXT_PROMPTS.get(
            self.class_id,
            f"a realistic photo of a {self.class_label}",
        )
        negative_prompt = (
            "blurry, low quality, distorted, artifacts, bad anatomy, low resolution"
        )

        synthetic_samples: List[Image.Image] = []

        for _ in tqdm(range(num_images), desc=f"Generating {self.class_label} samples"):
            try:
                if config.ENABLE_CONTROLNET and isinstance(
                    self.pipeline, StableDiffusionControlNetPipeline
                ):
                    control_image = self._build_control_image((512, 512))
                    result = self.pipeline(
                        prompt=prompt,
                        negative_prompt=negative_prompt,
                        image=control_image,
                        num_inference_steps=config.DIFFUSION_STEPS,
                        guidance_scale=config.CFG_GUIDANCE_SCALE,
                        height=256,
                        width=256,
                    )
                else:
                    result = self.pipeline(
                        prompt=prompt,
                        negative_prompt=negative_prompt,
                        num_inference_steps=config.DIFFUSION_STEPS,
                        guidance_scale=config.CFG_GUIDANCE_SCALE,
                        height=256,
                        width=256,
                    )

                image = result.images
                image = image.resize((32, 32), Image.Resampling.LANCZOS)
                synthetic_samples.append(image)
            except Exception as exc:
                print(f"[Gen] Generation error: {exc}")
                continue

        return synthetic_samples

    def dump_images(self, images: List[Image.Image], output_dir: Path) -> None:
        output_dir.mkdir(parents=True, exist_ok=True)

        for idx, image in enumerate(images):
            save_path = output_dir / f"{self.class_label}_{idx:04d}.png"
            image.save(save_path)

        print(f"[Gen] Saved {len(images)} synthetic images to {output_dir}")


def build_synthetic_dataset() -> List[Image.Image]:
    minority_generator = SyntheticImageGenerator(
        class_id=config.MINORITY_CLASS_ID,
        class_label="cat",
    )

    synthetic_batch = minority_generator.generate_batch(
        num_images=config.NUM_SYNTH_IMAGES,
    )

    minority_generator.dump_images(
        images=synthetic_batch,
        output_dir=config.SYNTH_IMAGES_DIR,
    )

    return synthetic_batch
```

Model

```python
def build_classifier(backbone_name: str = "ResNet50") -> nn.Module:
    if backbone_name == "ResNet50":
        model = models.resnet50(pretrained=False)  
        model.fc = nn.Linear(model.fc.in_features, config.NUM_LABELS)

    elif backbone_name == "ResNet18":
        model = models.resnet18(pretrained=False) 
        model.fc = nn.Linear(model.fc.in_features, config.NUM_LABELS)

    elif backbone_name == "ViT":
        try:
            import timm
            model = timm.create_model(
                "vit_base_patch16_224",
                pretrained=False,
                num_classes=config.NUM_LABELS,
            )  # standard ViT usage pattern[3]
        except ImportError:
            print("[Model] 'timm' is not installed, falling back to ResNet50 backbone")
            model = models.resnet50(pretrained=False)
            model.fc = nn.Linear(model.fc.in_features, config.NUM_LABELS)

    else:
        raise ValueError(f"Unsupported backbone: {backbone_name}")

    return model
```

Trainer

```python
class Trainer:
    def __init__(self, model: nn.Module, device: torch.device):
        self.model = model.to(device)
        self.device = device

        self.criterion = nn.CrossEntropyLoss()
        self.optimizer = optim.AdamW(
            self.model.parameters(),
            lr=config.BASE_LR,
            weight_decay=config.L2_WEIGHT_DECAY,
        )
        self.scheduler = CosineAnnealingLR(
            self.optimizer,
            T_max=config.TRAIN_EPOCHS,
            eta_min=1e-6,
        )

        self.history = {
            "train_loss": [],
            "val_acc": [],
            "val_balanced_acc": [],
        }

    def train_one_epoch(self, train_loader: DataLoader) -> float:
        self.model.train()
        running_loss = 0.0

        for images, targets in tqdm(train_loader, desc="Training"):
            images = images.to(self.device, non_blocking=True)
            targets = targets.to(self.device, non_blocking=True)

            self.optimizer.zero_grad()

            logits = self.model(images)
            loss = self.criterion(logits, targets)

            loss.backward()
            self.optimizer.step()

            running_loss += loss.item()

        epoch_loss = running_loss / len(train_loader)
        self.history["train_loss"].append(epoch_loss)
        return epoch_loss

    def evaluate(self, data_loader: DataLoader) -> Tuple[float, float]:
        self.model.eval()
        num_correct = 0
        num_samples = 0
        all_predictions: list[int] = []
        all_targets: list[int] = []

        with torch.no_grad():
            for images, targets in tqdm(data_loader, desc="Evaluating"):
                images = images.to(self.device, non_blocking=True)
                targets = targets.to(self.device, non_blocking=True)

                logits = self.model(images)
                _, preds = torch.max(logits, dim=1)

                num_samples += targets.size(0)
                num_correct += (preds == targets).sum().item()

                all_predictions.extend(preds.cpu().numpy().tolist())
                all_targets.extend(targets.cpu().numpy().tolist())

        top1_accuracy = 100.0 * num_correct / num_samples
        balanced_top1 = 100.0 * balanced_accuracy_score(all_targets, all_predictions)

        self.history["val_acc"].append(top1_accuracy)
        self.history["val_balanced_acc"].append(balanced_top1)

        return top1_accuracy, balanced_top1

    def fit(self, train_loader: DataLoader, val_loader: DataLoader, num_epochs: int):
        print(f"[Train] Starting training for {num_epochs} epochs on {self.device}...")

        for epoch in range(num_epochs):
            train_loss = self.train_one_epoch(train_loader)
            val_acc, val_balanced = self.evaluate(val_loader)

            self.scheduler.step()

            if (epoch + 1) % 10 == 0 or epoch == 0 or epoch + 1 == num_epochs:
                print(
                    f"[Epoch {epoch + 1}/{num_epochs}] "
                    f"Loss: {train_loss:.4f} | "
                    f"Acc: {val_acc:.2f}% | "
                    f"Balanced Acc: {val_balanced:.2f}%"
                )

        return self.history
```

Train

```python
class AugmentedCIFAR10Dataset(Dataset):
    def __init__(
        self,
        base_dataset: Dataset,
        synthetic_root: Path,
        synthetic_class_id: int,
        transform=None,
    ):
        self.base_dataset = base_dataset
        self.transform = transform
        self.synthetic_class_id = synthetic_class_id
        self.synthetic_images: list[Image.Image] = []
        self.synthetic_labels: list[int] = []

        self._load_synthetic_batch(synthetic_root)

    def _load_synthetic_batch(self, synthetic_root: Path) -> None:
        if not synthetic_root.exists():
            print(f"[Dataset] Synthetic directory not found: {synthetic_root}")
            return

        for img_path in synthetic_root.glob("*.png"):
            try:
                img = Image.open(img_path).convert("RGB")
                self.synthetic_images.append(img)
                self.synthetic_labels.append(self.synthetic_class_id)
            except Exception as exc:
                print(f"[Dataset] Failed to load {img_path}: {exc}")

        print(
            f"[Dataset] Loaded {len(self.synthetic_images)} synthetic samples "
            f"for class id {self.synthetic_class_id}"
        )

    def __len__(self) -> int:
        return len(self.base_dataset) + len(self.synthetic_images)

    def __getitem__(self, index: int):
        if index < len(self.base_dataset):
            image, label = self.base_dataset[index]
        else:
            synthetic_index = index - len(self.base_dataset)
            image = self.synthetic_images[synthetic_index]
            label = self.synthetic_labels[synthetic_index]

        if self.transform is not None:
            image = self.transform(image)

        return image, label


def run_complete_pipeline():
    print("=" * 80)
    print("EXPERIMENT: SYNTHETIC CIFAR-10 DATA WITH STABLE DIFFUSION + CONTROLNET")
    print("=" * 80)

    print("\n Building CIFAR-10 data loaders...")[1]
    data_loaders = build_data_loaders()

    print("\n Generating synthetic minority-class samples (may require substantial GPU memory)...")[2]
    try:
        build_synthetic_dataset()
        print("[OK] Synthetic samples generated successfully.")
    except Exception as exc:
        print(f"[WARNING] Failed to generate new synthetic data: {exc}")
        print("[INFO] Continuing with any pre-generated synthetic samples in the output directory.")

    transform_train = transforms.Compose([
        transforms.RandomCrop(32, padding=4),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
        transforms.Normalize(
            mean=(0.4914, 0.4822, 0.4465),
            std=(0.2470, 0.2430, 0.2610),
        ),
    ])

    transform_eval = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(
            mean=(0.4914, 0.4822, 0.4465),
            std=(0.2470, 0.2430, 0.2610),
        ),
    ])

    print("\n Scenario 1: Training WITHOUT synthetic data...")[3]
    baseline_model = build_classifier(config.BACKBONE_TYPE)
    baseline_trainer = Trainer(baseline_model, config.COMPUTE_DEVICE)

    imbalanced_train_loader = DataLoader(
        data_loaders["train_imbalanced"],
        batch_size=config.TRAIN_BATCH_SIZE,
        shuffle=True,
        num_workers=8,
        pin_memory=True,
    )

    baseline_history = baseline_trainer.fit(
        train_loader=imbalanced_train_loader,
        val_loader=data_loaders["test"],
        num_epochs=config.TRAIN_EPOCHS,
    )

    print("\n Scenario 2: Training WITH synthetic minority-class data...")[4]
    augmented_train_dataset = AugmentedCIFAR10Dataset(
        base_dataset=data_loaders["train_imbalanced_dataset"]
        if "train_imbalanced_dataset" in data_loaders
        else data_loaders["train_full_dataset"],
        synthetic_root=config.SYNTH_IMAGES_DIR,
        synthetic_class_id=config.MINORITY_CLASS_ID,
        transform=transform_train,
    )

    augmented_train_loader = DataLoader(
        augmented_train_dataset,
        batch_size=config.TRAIN_BATCH_SIZE,
        shuffle=True,
        num_workers=8,
        pin_memory=True,
    )

    synthetic_model = build_classifier(config.BACKBONE_TYPE)
    synthetic_trainer = Trainer(synthetic_model, config.COMPUTE_DEVICE)

    synthetic_history = synthetic_trainer.fit(
        train_loader=augmented_train_loader,
        val_loader=data_loaders["test"],
        num_epochs=config.TRAIN_EPOCHS,
    )

    print("\n" + "=" * 80)
    print("ABLATION STUDY RESULTS")
    print("=" * 80)

    results = {
        "without_synthetic": {
            "final_accuracy": baseline_history["val_acc"][-1],
            "final_balanced_acc": baseline_history["val_balanced_acc"][-1],
            "best_accuracy": max(baseline_history["val_acc"]),
            "best_balanced_acc": max(baseline_history["val_balanced_acc"]),
            "avg_train_loss": float(np.mean(baseline_history["train_loss"][-10:])),
        },
        "with_synthetic": {
            "final_accuracy": synthetic_history["val_acc"][-1],
            "final_balanced_acc": synthetic_history["val_balanced_acc"][-1],
            "best_accuracy": max(synthetic_history["val_acc"]),
            "best_balanced_acc": max(synthetic_history["val_balanced_acc"]),
            "avg_train_loss": float(np.mean(synthetic_history["train_loss"][-10:])),
        },
    }

    print("\n{:<30} {:<20} {:<20}".format("Metric", "Without synthetic", "With synthetic"))
    print("-" * 70)

    for metric in ["final_accuracy", "final_balanced_acc", "best_accuracy", "best_balanced_acc"]:
        base_value = results["without_synthetic"][metric]
        synth_value = results["with_synthetic"][metric]
        improvement = synth_value - base_value
        print(f"{metric:<30} {base_value:<20.2f} {synth_value:<20.2f} (+{improvement:.2f})")

    print("\nAverage training loss over the last 10 epochs:")
    print(f"  Without synthetic: {results['without_synthetic']['avg_train_loss']:.4f}")
    print(f"  With synthetic:    {results['with_synthetic']['avg_train_loss']:.4f}")

    config.RESULTS_DIR.mkdir(parents=True, exist_ok=True)
    with open(config.RESULTS_DIR / "ablation_results.json", "w") as fp:
        json.dump(results, fp, indent=2)

    plot_training_curves(baseline_history, synthetic_history)

    return results


def plot_training_curves(history_baseline, history_synthetic):
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    axes.plot(
        history_baseline["val_acc"],
        label="Without Synthetic",
        marker="o",
        markersize=3,
        alpha=0.7,
    )
    axes.plot(
        history_synthetic["val_acc"],
        label="With Synthetic",
        marker="s",
        markersize=3,
        alpha=0.7,
    )
    axes.set_xlabel("Epoch")
    axes.set_ylabel("Accuracy (%)")
    axes.set_title("Validation Accuracy")
    axes.legend()
    axes.grid(True, alpha=0.3)

    axes.plot([1]
        history_baseline["val_balanced_acc"],
        label="Without Synthetic",
        marker="o",
        markersize=3,
        alpha=0.7,
    )
    axes.plot([1]
        history_synthetic["val_balanced_acc"],
        label="With Synthetic",
        marker="s",
        markersize=3,
        alpha=0.7,
    )
    axes.set_xlabel("Epoch")[1]
    axes.set_ylabel("Balanced Accuracy (%)")[1]
    axes.set_title("Validation Balanced Accuracy")[1]
    axes.legend()[1]
    axes.grid(True, alpha=0.3)[1]

    plt.tight_layout()
    output_path = config.RESULTS_DIR / "training_curves.png"
    plt.savefig(output_path, dpi=150, bbox_inches="tight")
    print(f"\n[Plots] Saved training curves to {output_path}")
    plt.close()


if __name__ == "__main__":
    print(f"[INFO] CUDA available: {torch.cuda.is_available()}")
    if torch.cuda.is_available():
        print(f"[INFO] GPU: {torch.cuda.get_device_name(0)}")
        print(
            f"[INFO] GPU memory: "
            f"{torch.cuda.get_device_properties(0).total_memory / 1e9:.0f} GB"
        )

    experiment_results = run_complete_pipeline()

    print("\n[COMPLETE] Experiment finished.")
    print(f"[INFO] Results written to: {config.RESULTS_DIR}")
```
